```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
from keras.callbacks import EarlyStopping
```

```python
sentences = [
    "I love machine learning",
    "Machine learning is very useful",
    "I love natural language processing"
]

words = []

for sentence in sentences:
    for word in sentence.lower().split():
        if word not in words:
            words.append(word)

print("Words:", words)
print("Total Words:", len(words))
```

```python
word_to_id = {} 

for i in range(len(words)):
    word_to_id[words[i]] = i

print("Word Mapping:", word_to_id)

X = []
y = []

for sentence in sentences:

    sentence_words = sentence.lower().split()

    for i in range(1, len(sentence_words) - 1):

        context = [
            word_to_id[sentence_words[i - 1]],
            word_to_id[sentence_words[i + 1]]
        ]

        target = word_to_id[sentence_words[i]]

        X.append(context)
        y.append(target)

X = np.array(X)
y = np.array(y)

print("Input (Context):", X)
print("Target:", y)
```

```python
model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Embedding(len(words), 8),
    layers.GlobalAveragePooling1D(),
    layers.Dense(len(words), activation="softmax")
])

early_stop = EarlyStopping(monitor='loss',patience=7,restore_best_weights=True)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

model.fit(X, y, epochs=200,callbacks=[early_stop], verbose=1)
```

```python
cw1=input()
cw2=input()

test = np.array([[
    word_to_id[cw1],
    word_to_id[cw2]
]])

result = model.predict(test, verbose=0)

predicted_id = np.argmax(result)

print(f"Context Words: {cw1}, {cw2}")
print("Predicted Word:", words[predicted_id])
```
